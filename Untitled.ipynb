```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
plt.style.use('dark_background')
```

```python
df=pd.read_csv(r"C:\Users\vs333\Downloads\archive\zomato.csv")
df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
df.drop_duplicates(inplace=True)
df.shape
```

```python
df = df.drop(
    ['url','address','phone','menu_item','dish_liked','reviews_list'],
    axis=1,
    errors='ignore'
)

df.head()
```

```python
df['rate'].unique()
```

```python
def handlerate(value):
    if(value=='NEW' or value=='-'):
        return np.nan
    else:
        value= str(value).split('/')
        value = value[0]
        return float(value)

df['rate'] = df['rate'].apply(handlerate)
df['rate'].head()
```

```python
df.dropna(inplace = True)
df.head()
```

```python
df.rename(columns = {'approx_cost(for two people)':'coast2plates','listed_in(type)':'Type'},inplace=True)
df.head()
```

```python
df=df.drop(['listed_in(city)'],axis=1)
```

```python
df.head()
```

```python
df['coast2plates'].unique()
```

```python
def handlecomma(value):
    value = str(value)
    if ',' in value:
        value = value.replace(',','')
        return float(value)
    else:
        return float(value)
df['coast2plates'] = df['coast2plates'].apply(handlecomma)
df['coast2plates'].unique()
```

```python
df.head()
```

```python
rest_types = df['rest_type'].value_counts(ascending = False)
rest_types
```

```python
rest_types_lessthan1000 = rest_types[rest_types<1000]
rest_types_lessthan1000
```

```python
def handle_rest_type(value):
 if(value in rest_types_lessthan1000):
     return 'others'
 else:
     return value
df['rest_type'] = df['rest_type'].apply(handle_rest_type)
df['rest_type'].value_counts()
```

```python
location = df['location'].value_counts(ascending = False)
location_lessthan300 = location[location<300]

def handle_location(value):
    if(value in location_lessthan300):
        return 'others'
    else:
        return value
df['location'] = df['location'].apply(handle_location)
df['location'].value_counts()
```

```python
cuisines = df['cuisines'].value_counts(ascending = False)
cuisines_lessthan100 = cuisines[cuisines<100]

def handle_cuisines(value):
    if(value in cuisines_lessthan100):
        return 'others'
    else:
        return value
df['cuisines'] = df['cuisines'].apply(handle_cuisines)
df['cuisines'].value_counts()
```
